```python
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import FuncFormatter

# Calculate sales by category
category_sales = df.groupby('Category')['Amount'].sum().sort_values(ascending=False)

# Create figure
plt.figure(figsize=(12, 7))

ax = sns.barplot(
    x=category_sales.index,
    y=category_sales.values,
    hue=category_sales.index,
    palette='viridis',
    legend=False
)

# Add title and labels
plt.title('Sales Distribution by Product Category', fontsize=16, fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Sales Amount (₹)', fontsize=12)

# Convert Y-axis values into lakhs
ax.yaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f'₹{x/100000:.0f}L')
)

# Add values above bars
for bar in ax.patches:
    height = bar.get_height()
    if height > 0:
        ax.annotate(
            f'₹{height/100000:.1f}L',
            (bar.get_x() + bar.get_width()/2, height),
            ha='center',
            va='bottom',
            fontsize=9
        )

plt.xticks(rotation=35, ha='right')
plt.tight_layout()
plt.show()
```

```python
df['Date'] = pd.to_datetime(df['Date'])

monthly_sales = df.groupby(df['Date'].dt.month)['Amount'].sum()

plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales Amount (₹)')
plt.xticks(range(1, 13))

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Status')

plt.title('Order Status Distribution')
plt.xlabel('Order Status')
plt.ylabel('Number of Orders')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_categories = df['Category'].value_counts().head(10)

plt.figure(figsize=(8, 5))
sns.barplot(x=top_categories.values, y=top_categories.index)

plt.title('Top 10 Categories by Number of Orders')
plt.xlabel('Number of Orders')
plt.ylabel('Category')

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Fulfilment')

plt.title('Orders by Fulfilment Method')
plt.xlabel('Fulfilment Method')
plt.ylabel('Number of Orders')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Size', order=df['Size'].value_counts().index)

plt.title('Order Distribution by Size')
plt.xlabel('Size')
plt.ylabel('Number of Orders')

plt.show()
```

```python
top_states = df['ship-state'].value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_states.values, y=top_states.index)

plt.title('Top 10 States by Number of Orders')
plt.xlabel('Number of Orders')
plt.ylabel('State')

plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.countplot(data=df, x='B2B')

plt.title('B2B vs B2C Orders')
plt.xlabel('Business Order')
plt.ylabel('Number of Orders')

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Courier Status')

plt.title('Courier Status Distribution')
plt.xlabel('Courier Status')
plt.ylabel('Number of Orders')

plt.show()
```

```python
Q1 = df['Amount'].quantile(0.25)
Q3 = df['Amount'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Amount'] < lower_bound) | (df['Amount'] > upper_bound)]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(x=df['Amount'])

plt.title('Outlier Analysis of Sales Amount')
plt.xlabel('Amount (₹)')

plt.show()
```
